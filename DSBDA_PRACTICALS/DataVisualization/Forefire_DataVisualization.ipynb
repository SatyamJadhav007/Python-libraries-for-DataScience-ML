```python
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns 
import numpy as np
sns.set(style="whitegrid")
```

```python
df=pd.read_csv('forestfires.csv') 
df
```

```python
df.info()
```

```python
df=df.drop(columns=['X','Y','day']) 
df
```

```python
month_order=['jan','feb','mar','apr','may','jun','jul','aug','sep','oct','nov','dec'] 

df['month']=pd.Categorical(df['month'],categories=month_order,ordered=True) 
df
```

```python
df['log_area'] = np.log1p(df['area']) #Used for c) for better visualiztion of data (decreasing skewness)
#np.log1p(df['area']) 
df
```

```python
# a) Line Graph for burned area over the months
monthly_area=df.groupby('month')['area'].sum().reset_index()
monthly_area
```

```python
plt.figure(figsize=(10,5)) 
sns.lineplot(data=monthly_area,x='month',y='area',marker='o') 

plt.title('Monthly Variation in Burned Area') 
plt.xlabel('Month') 
plt.ylabel('Total Burned Area') 
plt.xticks(rotation=45) 
plt.show()
```

```python
# b.) Bar Chart  for Fire Frequency across months 
# Number of  Fire Incidents  per month
monthly_freq=df['month'].value_counts().sort_index().reset_index() 
# monthly_freq.columns=['month','count']
```

```python
plt.figure(figsize=(10,5)) 
sns.barplot(data=monthly_freq,x="month",y='count') 

plt.title("Fire Frequency Across Months") 
plt.xlabel("Month")  
plt.ylabel("Number of Fires") 
plt.xticks(rotation=45)
plt.show()
```

```python
# c.) Histogram -> Distribution of  Burned Area 
plt.figure(figsize=(10,5)) 
sns.histplot(df['log_area'],bins=30,kde=True)  

plt.title("Distuibution of Burned Area") 
plt.xlabel("Log(Area + 1)") 
plt.ylabel("Frequency") 
plt.show() 
```

```python
# d.) Scatter plot ->wind Speed vs Fire Spread Area 

plt.figure(figsize=(10,5)) 
sns.scatterplot(data=df,x='wind',y='log_area',alpha=0.6)  

plt.title("Wind Speed vs Fire Spread Area") 
plt.xlabel("Wind Speed") 
plt.ylabel("Log(Area + 1)") 
plt.show() 
```

```python
# e.) Temperature Impact on Fire occurrence (Box Plot)
df['temp_bin']=pd.cut(df['temp'],bins=5)  

plt.figure(figsize=(12,6)) 
sns.boxplot(data=df,x='temp_bin',y='log_area') 

plt.title("Temperature Impact on Fire Spread") 
plt.xlabel("Temperature Range") 
plt.ylabel("Burned Area")  
plt.xticks(rotation=45)
plt.show()
```

```python
# f.) heatmap correlation between different environmental factors
corr_data = df[['temp', 'RH', 'wind', 'rain']]
# Focus on temp,rh,wind,rain 
corr_matrix = corr_data.corr()

plt.figure(figsize=(8,6))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt=".2f")

plt.title("Correlation Heatmap of Environmental Factors")
plt.show()
```
