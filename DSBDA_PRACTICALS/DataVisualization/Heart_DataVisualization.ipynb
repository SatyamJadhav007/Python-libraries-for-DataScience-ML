```python
import pandas as pd 
import numpy as np  
import matplotlib.pyplot as plt 
import seaborn as sns 
sns.set(style='whitegrid')
plt.rcParams['figure.figsize']=(8,5)
```

```python
df['sex'] = df['sex'].map({0: 'Female', 1: 'Male'})
df['output'] = df['output'].map({0: 'No Disease', 1: 'Disease'})

# (Optional but useful)
df['exng'] = df['exng'].map({0: 'No', 1: 'Yes'})
df['fbs'] = df['fbs'].map({0: 'Normal', 1: 'High'}) 
```

```python
# a.)CountPlot Disease or not distribution
sns.countplot(x='output',data=df) 
plt.title("Distribution of Heart Disease")  
plt.xlabel("Heart Disease Status") 
plt.ylabel("Count")  
plt.show()
```

```python
# b.) Bar Chart (Heart Disease across gender)  

sns.countplot(x='sex',hue='output',data=df) 
plt.title("Heart Disease Cases Across Gender") 
plt.xlabel("Gender") 
plt.ylabel("Count") 
plt.legend(title='Condition') 
plt.show()
```

```python
#c.) historgram distribtion for age,cholestrol,and resting BP

sns.histplot(data=df[['age']],bins=10,kde=True) 
plt.title("Age Distribution") 
plt.xlabel("age") 
plt.ylabel("Number of patients")
plt.show()

# Cholesterol
sns.histplot(df['chol'], kde=True)
plt.title("Cholesterol Distribution")
plt.xlabel("Cholesterol")
plt.ylabel("Frequency")
plt.show()

# Resting Blood Pressure
sns.histplot(df['trtbps'], kde=True)
plt.title("Resting Blood Pressure Distribution")
plt.xlabel("Resting BP")
plt.ylabel("Frequency")
plt.show()
```

```python
# d.) scatterplot for agevs heart rate 
plt.figure(figsize=(10,5)) 
sns.scatterplot(data=df,x='age',y='thalachh',hue='output') 
plt.title("Maximum Heart Rate vs Age")
plt.xlabel("Age")
plt.ylabel("Max Heart Rate")
plt.legend(title="Condition")
plt.show()
```

```python
# e) Box plot for cholesterol across heart disease classes
sns.boxplot(x='output', y='chol', data=df)

plt.title("Cholesterol Levels by Heart Disease Status")
plt.xlabel("Heart Disease")
plt.ylabel("Cholesterol")

plt.show()
```

```python
# f) HeatMap Dataset Correlaton 
df_corr=df.copy() 

df_corr['sex'] = df_corr['sex'].map({'Female': 0, 'Male': 1})
df_corr['output'] = df_corr['output'].map({'No Disease': 0, 'Disease': 1})
df_corr['exng'] = df_corr['exng'].map({'No': 0, 'Yes': 1})
df_corr['fbs'] = df_corr['fbs'].map({'Normal': 0, 'High': 1})  

corr=df_corr.corr() 

sns.heatmap(corr,annot=True,cmap='coolwarm',fmt='.2f') 

plt.title("Correlation Heatmap of Features") 
plt.show()
```
